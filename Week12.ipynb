# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

1. Side-by-Side Small Multiples (Figure 1: Two Separate Panels)

    Advantages:

        Clarity of Individual Shapes: Each week's trajectory is plotted on its own without visual interference or overlapping lines.

        No Clutter: Great for presenting data where two time series cross frequently or have vastly different scales (though here they share a scale).

    Disadvantages:

        Harder Day-by-Day Comparison: The viewer's eyes must jump back and forth across the white space to compare specific days (e.g., comparing Day 2 of Week 1 vs. Day 2 of Week 2 requires spatial memory).

        Redundant Axes: Consumes more horizontal space and duplicates the x-axis scale.

2. Overlaid Single Chart (Figure 2: Both Lines on One Plot)

    Advantages:

        Direct & Direct Magnitude Comparison: Makes it instantly obvious which week had higher or lower visitors on any exact day.

        Efficient Space: Compact and immediately highlights convergence, divergence, and gaps between the two weeks.

    Disadvantages:

        Potential Clutter/Spaghetti Graphing: If lines cross repeatedly or if you add more than 2–3 weeks, the chart quickly becomes unreadable.

        Missing Legend/Distinction: Without proper formatting (like color coding, labels, or linestyles), it is impossible to know which line represents Week 1 and which represents Week 2.

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt
import numpy as np

# Set up data (simulated 5-year trend representing Chapter 8 style redesigns)
years = ['2011', '2012', '2013', '2014', '2015']
baseline_metric = [100, 98, 95, 93, 90]
target_metric   = [80,  85, 90, 98, 115]

# Create figure with Storytelling With Data minimal dimensions
fig, ax = plt.subplots(figsize=(8, 5))

# Plot lines: Grey for background/context, bold primary color for the story focus
ax.plot(years, baseline_metric, color='#B0B0B0', linewidth=2.5, marker='o', label='Baseline / Competitor')
ax.plot(years, target_metric,   color='#005B94', linewidth=3.5, marker='o', label='Our Performance')

# Direct labeling at the end of the lines (avoiding separate legend box)
ax.text(years[-1], baseline_metric[-1] - 2, ' Competitor Average', color='#707070', fontsize=11, va='center')
ax.text(years[-1], target_metric[-1], ' Our Performance', color='#005B94', fontsize=12, fontweight='bold', va='center')

# Highlight key data points directly
for i, txt in enumerate(target_metric):
    ax.annotate(f'{txt}', (years[i], target_metric[i]), 
                textcoords="offset points", xytext=(0,10), 
                ha='center', fontweight='bold', color='#005B94')

# Storytelling With Data aesthetic cleanup (remove top/right borders, clean grid)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_color('#CCCCCC')
ax.spines['bottom'].set_color('#CCCCCC')

# Adjust ticks and labels
ax.tick_params(axis='both', which='both', length=0, labelsize=11, colors='#555555')
ax.set_ylim(60, 130)

# Clear action title & subtitle
plt.suptitle("Performance surpassed the competitor average by 2014", 
             x=0.08, y=0.98, ha='left', fontsize=14, fontweight='bold', color='#333333')
ax.set_title("Annual customer satisfaction index (2011–2015)", 
             loc='left', fontsize=11, color='#707070', pad=15)

plt.tight_layout()
plt.show()
```
