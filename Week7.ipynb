# 2. Weekly graph question

Below are the graphs from last week's weekly graph question.  Suppose you wanted to cause the viewer to examine the last three months of data first.  How could you use preattentive attributes to do that?  Suggest several possibilities.

```python
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size = numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Nitrate level (% of eutrophic value)")
```

```python
np.random.seed(1)
time_series = 100 * (1 - (np.random.normal(size = numdates) / 8 + np.arange(numdates) / 16))
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Phosphate level (% of eutrophic value)")
```

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# 1. Setup the data
numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size=numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='ME')

# 2. Create the base plot
plt.figure(figsize=(10, 6))
# Plot the full line in a muted color
sns.lineplot(x=datearray, y=time_series, color='lightgray', label='Historical')

# 3. Apply preattentive attributes to the last 3 months
# Subset the last 3 data points
last_three_dates = datearray[-3:]
last_three_series = time_series[-3:]

# Plot the highlighted segment in a high-contrast color and thicker line
sns.lineplot(x=last_three_dates, y=last_three_series, 
             color='red', linewidth=3, label='Recent Focus')

# Final styling
plt.xlabel("Year and month")
plt.ylabel("Nitrate level (% of eutrophic value)")
plt.legend()
plt.show()
```
